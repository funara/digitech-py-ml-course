## 📄 BAGIAN 2: Template SOP Produk (Standard Operating Procedure)

SOP berfungsi sebagai panduan operasional agar pengguna umum (non-teknis) atau penguji dapat menjalankan aplikasi Anda tanpa kendala.

> ✏️ **TUGAS TIM**: Salin template di bawah ini ke file Markdown terpisah (misal `SOP_Aplikasi.md`) atau isi langsung di dokumentoyek tim Anda.

### 1. Prasyarat Sistem (*System Prerequisites*)
- Browser modern (Google Chrome, Mozilla Firefox, Edge, Safari).
- Koneksi internet aktif (jika diakses via URL Streamlit Cloud).
- Python 3.9+ (jika dijalankan secara lokal).

### 2. Langkah Menjalankan Aplikasi Secara Lokal
1. Clone repositori Git proyek Anda:
   ```bash
   git clone <URL_REPOS_ANDA>
   cd <NAMA_FOLDER_REPOS>
   ```
2. Buat dan aktifkan *Virtual Environment* (opsional tetapi direkomendasikan):
   ```bash
   python -m venv venv
   # Windows
   .\\venv\\Scripts\\activate
   # macOS/Linux
   source venv/bin/activate
   ```
3. Install seluruh dependensi yang diperlukan:
   ```bash
   pip install -r requirements.txt
   ```
4. Jalankan aplikasi Streamlit:
   ```bash
   streamlit run Home.py
   ```

### 3. Panduan Penggunaan Fitur Utama
- **Halaman 1: Dashboard EDA (Exploratory Data Analysis)**
  - Masuk ke menu sidebar, pilih **EDA**.
  - Gunakan filter interaktif (multiselect/dropdown) untuk memfilter visualisasi berdasarkan kategori tertentu.
- **Halaman 2: Prediksi / Inference**
  - Pilih menu **Prediksi**.
  - Masukkan parameter input data pada form yang disediakan.
  - Klik tombol **"Prediksi Sekarang"** untuk memicu inferensi model ML.
  - Hasil prediksi beserta probabilitasnya (jika klasifikasi) akan ditampilkan di layar.

## 📖 BAGIAN 3: Template Glossary Istilah ML

Glosarium sangat membantu klien/penguji non-teknis untuk memahami istilah-istilah metrik evaluasi atau parameter ML yang muncul di dasbor Anda.

| Istilah | Definisi Singkat | Relevansi dalam Aplikasi |
| :--- | :--- | :--- |
| **EDA** | Proses analisis data awal untuk memahami pola dan distribusi data menggunakan visualisasi statistika. | Digunakan pada halaman analisis deskriptif data historis. |
| **Accuracy** | Rasio prediksi benar dibanding total keseluruhan data uji. | Metrik utama keandalan model klasifikasi. |
| **R² (R-squared)** | Nilai yang mengukur seberapa baik model regresi menjelaskan variabilitas target (skala 0-1). | Metrik utama untuk prediksi numerik. |
| **Feature Scaling** | Metode penyeragaman rentang nilai fitur numerik (misal normalisasi 0-1 atau standarisasi). | Mencegah fitur bernilai besar mendominasi proses pembelajaran model. |
| **Feature Importance** | Skor yang menunjukkan tingkat kontribusi suatu fitur terhadap prediksi model. | Membantu pengguna memahami faktor apa yang paling memengaruhi hasil prediksi. |
